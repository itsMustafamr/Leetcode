```python
from typing import List
from collections import defaultdict

class Solution:
    def groupAnagrams(self, strs: List[str]) -> List[List[str]]:
        # Method 1: Using sorted strings as keys (O(n * m * log m) time)
        anagrams = defaultdict(list)
        
        for s in strs:
            # Sort the string and use it as a key
            sorted_str = ''.join(sorted(s))
            anagrams[sorted_str].append(s)
        
        return list(anagrams.values())
    
    def groupAnagramsOptimized(self, strs: List[str]) -> List[List[str]]:
        # Method 2: Using character count as keys (O(n * m) time)
        anagrams = defaultdict(list)
        
        for s in strs:
            # Create a count array for each character
            count = [0] * 26  # For lowercase letters a-z
            
            for char in s:
                count[ord(char) - ord('a')] += 1
            
            # Use tuple of counts as key
            key = tuple(count)
            anagrams[key].append(s)
        
        return list(anagrams.values())

# Example usage and explanation
def demonstrate_group_anagrams():
    solution = Solution()
    
    # Test case
    strs = ["eat", "tea", "tan", "ate", "nat", "bat"]
    
    print("Original strings:", strs)
    print("\nMethod 1 (Sorted):")
    result1 = solution.groupAnagrams(strs)
    print(result1)
    
    print("\nMethod 2 (Character Count):")
    result2 = solution.groupAnagramsOptimized(strs)
    print(result2)
    
    print("\nExplanation:")
    print("- 'eat', 'tea', 'ate' are anagrams (same characters, different order)")
    print("- 'tan', 'nat' are anagrams")
    print("- 'bat' is alone (no anagrams)")
    
    print("\nKey concepts:")
    print("1. Anagrams have the same character frequency")
    print("2. Method 1: Sort each string to get a common key")
    print("3. Method 2: Count characters to get a frequency key")
    print("4. Use defaultdict to avoid checking if key exists")

# Run the demonstration
demonstrate_group_anagrams()




```
